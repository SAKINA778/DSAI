```python
# Q1. Write a Python function count_vowels that accepts a string and returns the number of vowels a, e, i, o, u, A, E, I, O, U in that string.
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count

result = count_vowels("Welcome")
print(result, "vowels")
# - Example: 'Welcome'
# - Output: 3 vowels
```

```python
# Q2. Write a Python function display_student_info that accepts the following details about a student:

# - name (Keyword argument): The name of the student.
# - age (Keyword argument): The age of the student.
# - grade (Keyword argument): The grade the student is in.
# - school (Keyword argument): The name of the school the student attends.

def display_student_info(name, age, grade, school):
    print("Name:", name)
    print("Age:", age)
    print("Grade", grade)
    print("School:", school)

display_student_info(name="John Doe", age= 15, grade= "10th", school = "Greenwood High")

# Expected Output:

# - Name: John Doe
# - Age: 15
# - Grade: 10th
# - School: Greenwood High
```

```python
# Q3. Write a Python function display_prices that takes a product name and an arbitrary number of keyword arguments representing product prices.
# The function should display each product and its price.

def display_prices(product_name, **price):
    print("Product:", product_name)
    print()
    for store, price in price.items():
        print(f"{store}: ${price}")

display_prices("laptop", amazon = 800, ebay= 750, Walmart=780)
# Expected Output:

# Product: Laptop

# Amazon: $800

# eBay: $750

# Walmart: $780
```

```python
# Q5. Write a lambda function to calculate the delivery charges.
# An online store charges Rs. 150 for delivery when the order is below Rs. 2,000. Orders of Rs. 2,000 or more have free delivery.

dc = lambda order_amnt: 150 if order_amnt < 2000 else 0

print(dc(1500))
print(dc(2500))
```

```python
# Q6.

# Write a Python function calculate_rectangle that accepts the length and width of a rectangle as arguments. Inside this function, create two nested
# functions:

# - calculate_area: This function should calculate the area of the rectangle.
# - calculate_perimeter: This function should calculate the perimeter of the rectangle.
# - The main function should return both the area and the perimeter of the rectangle.

def calculate_rectangle(length, width):

    def calculate_area():
        return length * width

    def calculate_perimeter():
        return 2 * (length + width)

    area = calculate_area()
    perimeter = calculate_perimeter()
    return area, perimeter

area, perimeter = calculate_rectangle(5,3)
print(f"Area: {area}")
print(f"Perimeter: {perimeter}")
# Expected Output:

# Area: 15

# Perimeter: 16
```

```python
# Q8. Write a nested function to calculate the total hotel bill based on the number of nights.
# A hotel charges Rs. 5,000 per night. The customer must also pay 10% tax on the room cost

def calculate_hotel_bill(nights):
    per_night = 5000

    def calculate_cost():
        room_cost = per_night * nights
        tax = room_cost * 0.10
        total = room_cost + tax
        return room_cost, tax, total 
        
    room_cost, tax, total = calculate_cost()
    print("Room Cost:", room_cost)
    print("Tax:", tax)
    print("Total Bill:", total)

calculate_hotel_bill(3)
# Expected Output:
# Room Cost: 15000
# Tax: 1500.0
# Total Bill: 16500.0
```

```python
# Q9.User Login Validation
# A website checks a user's username and password before allowing access. Write a nested function to validate the login details

def user_login(username, password):

    def validate():
        valid_username = "admin"
        valid_password = "1234"
        return username == valid_username and password == valid_password

    if validate():
        return "Login Successful"
    else:
        return "Invalid username or password"

print(user_login("admin", "1234"))
print(user_login("user", "wrong"))
```

```python
# Q10.

# Write a Python function shopping_bill(price, quantity, discount=0) that uses nested functions:

# - An inner function calculate_total() → multiplies price * quantity.
# - An inner function apply_discount(total) → subtracts discount from total (but never below 0).

# The outer function should return the final bill amount.

def shopping_bill(price, quantity, discount=0):
    
    def calculate_total():
        return price * quantity
    
    def apply_discount(total):
        final = total - discount
        if final < 0:
            return 0
        return final
    
    total = calculate_total()
    final_bill = apply_discount(total)
    return final_bill

print(f"Final Bill = {shopping_bill(200, 3)} PKR (no discount)") 
print(f"Final Bill = {shopping_bill(200, 3, 100)} PKR (with discount)") 


# Expected Output:

# - Final Bill = 600 PKR (no discount)
# - Final Bill = 500 PKR (with discount)
```
