```python
#Q2. You are given a list that contains some duplicate items. Remove the duplicates by converting the list to a set. Final output should be in list

shopping_list = ["apple", "banana", "apple", "orange", "banana", "grape"]

unique_set = set(shopping_list)

#Final output should be in list
unique_list = list(unique_set)
print(unique_list)
#Expected Output: ['apple', 'grape', 'banana', 'orange']
```

```python
#Q3. You are managing a list of students enrolled in two different courses. You need to perform various set operations to understand the
#student enrollment. Task: Create two sets: one for students in "Course A" and one for students in "Course B". Find the students who are in 
#Course A" but not in "Course B" (difference). Find students who are only in one of the two courses
#(symmetric difference). 
course_a = {"John", "Alice", "Bob", "David"} 
course_b = {"Alice", "Eve", "Charlie", "David"}
#Using set methods

#• Find students who are in Course A but not in Course B (difference)
only_in_a = course_a.difference(course_b)
print("Students in Course A but not in Course B:", only_in_a)

#• Find students who are only in one of the two courses (symmetric difference)
only_one_course = course_a.symmetric_difference(course_b)
print("Students only in one of the two course", only_one_course)

#Expected Output:
#Students in Course A but not in Course B: {'Bob', 'John'}
#Students only in one of the two courses: {'Eve', 'John', 'Bob', 'Charlie'}
```

```python
#Q4. 
numbers = (10, 20, 30, 40, 50, 60, 70)

#Display the first 3 elements.
print(numbers[:3])
#Display the last 3 elements.
print(numbers[-3:])
#Display alternate elements.
print(numbers[::2])
```

```python
#Q5. Combine Two Sets
A = {"Ali", "Sara"}
B = {"Ahmed", "Zara"}

#Combine both sets into one set.

combined = A.union(B)
print(combined)
```

```python
#Q6. Set Comparison
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

#Find common elements.
print(A & B)
#Find elements only in Set A.
print(A - B)
#Find elements only in Set B.
print(B - A)
```

```python
#Q9. Given the tuple 
coordinates = (4, 5, 6)
#unpack the elements of the tuple into three variables x, y, and z.Then, calculate and print the sum of the variables.
x, y, z = coordinates
total = x + y + z
print("X =",x)
print("Y =",y)
print("Z =" ,z)
print("SUM", total)
```
